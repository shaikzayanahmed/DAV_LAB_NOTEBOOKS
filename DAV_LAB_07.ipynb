Simple and Multiple Linear Regression: Building models,checking assumptions,and evaluating using R^2 and RMSE


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_squared_error
```

```python
df=pd.read_csv('housing_clean.csv')
df.head()
```

```python
#Simple input Feature:
from re import X
X_simple=df[['Area']]
y=df['Price']
X_tr,X_te,y_tr,y_te=train_test_split(X_simple,y,test_size=0.2,random_state=42)
model_simple=LinearRegression()
model_simple.fit(X_tr,y_tr)
pred_simple=model_simple.predict(X_te)
print('SIMPLE REGRESSION')
print('R^2:',round(r2_score(y_te,pred_simple),3))
print('RMSE:',round(np.sqrt(mean_squared_error(y_te,pred_simple)),2))

```

```python
#Four input Feature:
from re import X
X_multi=df[['Area','Rooms','Age','Location_Score']]
y=df['Price']
X_tr,X_te,y_tr,y_te=train_test_split(X_multi,y,test_size=0.2,random_state=42)
model_multi=LinearRegression()
model_multi.fit(X_tr,y_tr)
pred_multi=model_multi.predict(X_te)
print('Multiple REGRESSION')
print('R^2:',round(r2_score(y_te,pred_multi),3))
print('RMSE:',round(np.sqrt(mean_squared_error(y_te,pred_multi)),2))
```

```python
residulas=y_te-pred_multi

plt.scatter(pred_multi,residulas)
plt.axhline(0,color='red',linestyle='--')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals (Actula-predictaded)')
plt.title('residual plot')
plt.show()
```
